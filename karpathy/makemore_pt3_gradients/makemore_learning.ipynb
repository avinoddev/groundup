```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
import random

g = torch.Generator().manual_seed(2147483647)
random.seed(42)
```

```python
names = open('names.txt').read().splitlines()
len(names)
```

```python
chars = sorted(list(set(''.join(names))))

c_to_i = {c:i+1 for i,c in enumerate(chars)}
c_to_i['.'] = 0

i_to_c = {i:c for c,i in c_to_i.items()}
i_to_c
```

```python
def build_dataset(words): 
    X, Y = [], []
    block_size = 3

    for n in words:
        context = [0] * 3

        for ch in n + '.':
            ix = c_to_i[ch]
            Y.append(ix)
            X.append(context)

            context = context[1:] + [ix]

    return X, Y

n1 = int(0.8 * len(names))
n2 = int(0.9 * len(names))

# Split dataset into 80% (training) / 10% (dev) / 10% (testing)

X_train, Y_train = build_dataset(names[:n1])
X_dev, Y_dev = build_dataset(names[n1:n2])
X_train, Y_train = build_dataset(names[n2:])
```
