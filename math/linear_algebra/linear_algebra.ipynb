```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([3, 2])            # 2d vector
v2 = np.array([-2, -4])         # 2d vector
v3 = np.array([-5])             # 1d vector
```

Vector: An ordered list of numbers

Below is an implementation of simple 2d and 1d vectors

```python
def plot_2d_vector(v):
    fig, ax = plt.subplots()

    color = np.random.rand(3,)  # random RGB color

    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=color)

    ax.set_xlim(-6, 6)
    ax.set_ylim(-6, 6)
    ax.set_xticks(np.arange(-6, 7, 1))
    ax.set_yticks(np.arange(-6, 7, 1))

    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.grid(True)

    plt.show()

plot_2d_vector(v)
```

Below is checking if a figure in linearly independent or linearly dependent

```python
def is_dependent(v1, v2):
    cross = v1[0]*v2[1] - v1[1]*v2[0]

    if cross == 0:
        print("Linearly dependent")
    else:
        print("Linearly independent")

is_dependent(v, v2)
```

Below is an implementation of plotting span

```python
def plot_span(v1, v2, num_iterations=3000, color=None):

    if color is None:
        color = np.random.rand(3,)

    fig, ax = plt.subplots()

    for i in range(num_iterations):
        a = np.random.uniform(-3, 3)
        b = np.random.uniform(-3, 3)

        point = a*v1 + b*v2
        ax.plot(point[0], point[1], '.',color=color, markersize=1)

    ax.set_xlim(-20, 20)
    ax.set_ylim(-20, 20)

    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)


    plt.show()

plot_span(v, v2, color='red')


    
```
